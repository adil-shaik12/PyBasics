```python
#tuples
tuple1=("apple","banana","orange","guava")
print(tuple1)
print(len(tuple1))
print(type(tuple1))
#update tuples: tuples are unchangeable, so to add tuples, we have to use another tuple and join them and update the original tuple
x=("apple", "banana", "cherry")
y=("orange",)
x+= y
print(x)
#remove items from tuple: tuples are unchangeable, so to remove items, we have to convert tuple into list, remove the item and convert back to tuple
myTuple=("apple","orange","banana","guava")
tupleList= list(myTuple)
tupleList.remove("apple")
myTuple=tuple(tupleList)
print(myTuple)
```
