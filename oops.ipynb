```python
#__init__()method
class lang:
    def __init__(self,name,pages):
        self.name=name
        self.pages=pages
lan1=lang("java",200)
lan2=lang("python",200)
print(lan1.name,lan1.pages)
print(lan2.name,lan2.pages)
#multiple parameters
class student:
    def __init__(self,name,age,course,rollno):
        self.name=name
        self.age=age
        self.course=course
        self.rollno=rollno
stu1=student("adil",20,"cse",1)
stu2=student("shaik",21,"ece",4)
print("name:",stu1.name,"age:",stu1.age,"course:",stu1.course,"rollno:",stu1.rollno)
print("name:",stu2.name,"age:",stu2.age,"course:",stu2.course,"rollno:",stu2.rollno)
#using self parameter
class student:
    def __init__(self,name,age,college,rollno):
        self.name=name
        self.age=age
        self.college=college
        self.rollno=rollno
    def person(self):
        print("my name is ",self.name)
        print("i am",self.age,"years old")
        print("i am studing in",self.college)
        print("my rollno is",self.rollno)
stu=student("mahammad",20,"cbit",12)
stu.person()
```

```python
#calling methods with self
class Person:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return "Hello, " + self.name
    def welcome(self):
        message = self.greet()
        print(message )
p1 = Person("adil")
p1.welcome()
#class properties
#add new properties
class lang:
    def __init__(self,name):
        self.name=name
lan1=lang("java")
lan2=lang("python")
lan1.pages=200
lan2.pages=250
print(lan1.name,lan1.pages)
print(lan2.name,lan2.pages)
#modify properties
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def myfunc(self):
        print("Hello my name is " +self.name)
p1=Person("adil", 20)
p1.myfunc()
p1.age=19
print(p1.age)
```
