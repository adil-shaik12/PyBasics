```python
items={"car","bike","cycle","car"}
print(items)
print(len(items))
print(type(items))
#accessing elements
for x in items:
    print(x)
#checking for an element
item1={"pulsar","yamaha","royalenfield"}
print("yamaha" in item1)
print("volvo" not in item1)
#adding elements
item1.add("activa")
print(item1)
item2=["unicorn","shine","passionpro"]
item1.update(item2)
print(item1)
#remove items
item1.remove("unicorn")
print(item1)
item1.discard("honda")
print(item1)
#using pop
x=item1.pop()
print(x)
print(item1)
#using clear
item1.clear()
print(item1)
```
