```python
#lists
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits)
print(len(fruits))
list2=[1,2,3,4]
list3=[True,True,False]
print(type(fruits))
print(type(list2))
print(type(list3))
#accessing lists
print(fruits[3])
print(fruits[-3])
print(fruits[0:3])
print(fruits[-5:-1])
print(fruits[0:])
print(fruits[-5:])
#change list item
fruits[3]="doll"
print(fruits)
fruits[3:5]=["guava","orange"]
print(fruits)
fruits.insert(3,"orange")
print(fruits)
#add items in list
fruits.append("orange")
print(fruits)
fruits1=["mango", "pineapple", "papaya"]
fruits.extend(fruits1)
print(fruits)
#remove list items
listA = ["apple", "banana", "cherry"]
listA.remove("banana")
print(listA)
listB = ["apple", "banana", "cherry"]
listB.pop(2)
print(listB)
listC = ["apple", "banana", "cherry"]
del listC[0]
print(listC)
listD = ["apple", "banana", "cherry"]
listD.clear()
print(listD)
```

```python
#loop in list
list0 = ["apple", "banana", "cherry"]
for x in list0:
    print(x)
#using while
i = 0
while i<len(list0):
    print(list0[i])
    i = i + 1
#sort
list1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
list1.sort()
print(list1)
#reverse
list2 = ["orange", "mango", "kiwi", "pineapple", "banana"]
list2.reverse()
print(list2)
#copy list
list3 = ["apple", "banana", "cherry"]
list1 = list3.copy()
print(list1)
#join lists
list4=["a", "b", "c"]
list5=["apple"]
print(list4+list5)
#joining lists using forloop
list6 = ["a", "b" , "c"]
list7 = [1, 2, 3]
for x in list7:
    list6.append(x)
    print(list6)
#using extend method
list8 = ["a", "b" , "c"]
list9 = [1, 2, 3]
list8.extend(list9)
print(list8)
```
