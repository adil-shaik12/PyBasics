```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.__age=age
    def get_age(self):
        return self.__age
p1=Person("adil", 20)
print(p1.name)
print(p1.get_age())
```

```python
class Person1:
    def __init__(self,name,age):
        self.name=name
        self.__age=age
    def get_age(self):
        return self.__age
    def set_age(self, age):
        if age > 0:
            self.__age=age
        else:
            print("Age must be positive")
p1 = Person1("shaik", 20)
print(p1.get_age())
p1.set_age(21)
print(p1.get_age())
```
