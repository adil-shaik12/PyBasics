```python
dict0={
    "brand":"toyota",
    "model":"fortuner",
    "year":2022
    }
print(dict0)
print(dict0["model"])
print(len(dict0))
print(type(dict0))
#accessing items
x=dict0["brand"]
print(x)
y=dict0.get("model")
print(y)
z=dict0.keys()
print(z)
a=dict0.values()
print(a)
b=dict0.items()
print(b)
#change items
dict0['year']=2020
print(dict0)
dict0.update({"year":2021})
print(dict0)
#adding items
dict0["color"]="black"
print(dict0)
dict0.update({"color":"white"})
print(dict0)
#removing items
dict0.pop("year")
print(dict0)
del dict0["color"]
print(dict0)
```

```python
#loops
dict1={
    "brand":"toyota",
    "model":"fortuner",
    "year":2022
    }
for x in dict1:
    print(x)
    print(dict1[x])
for x,y in dict1.items():
    print(x,y)
#coping dictionaries
dict2=dict1.copy()
print(dict2)
#nested dictionaries
library={
    "java":{
        "pages":200,
        "price":100
    },
    "python":{
        "pages":250,
        "price":150
    },
    "c++":{
        "pages":200,
        "price":150
    },
}
print(library)
#access items in nested dictionaries
print(library["java"]["price"])
#loop through nested dictionaries has to ask
```
