```python
class fruits:
    def __init__(self,ffruit,sfruit,tfruit):
        self.x=ffruit
        self.y=sfruit
        self.z=tfruit
    def print(self):
        print(self.x,self.y,self.z)
class fruits1(fruits):
    def __init__(self,ffruit,sfruit,tfruit):
        fruits.__init__(self,ffruit,sfruit,tfruit)
a=fruits1("apple","banana","mango")
a.print()
```
