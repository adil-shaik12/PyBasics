```python
#lists
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits)
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(len(fruits))
list1=["apple", "banana", "cherry", "apple", "cherry"]
list2=[1,2,3,4]
list3=[True,True,False]
print(type(list1))
print(type(list2))
print(type(list3))
#accessing lists
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[3])
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[-3])
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[0:3])
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[-5:-1])
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[0:])
fruits=["apple", "banana", "cherry", "apple", "cherry"]
print(fruits[-5:])
#change list item
fruits=["apple", "banana", "cherry", "apple", "cherry"]
fruits[3]="doll"
print(fruits)
fruits=["apple", "banana", "cherry", "apple", "cherry"]
fruits[3:5]=["guava","orange"]
print(fruits)
fruits=["apple", "banana", "cherry", "apple", "cherry"]
fruits.insert(3,"orange")
print(fruits)
#add items in list
fruits= ["apple", "banana", "cherry"]
fruits.append("orange")
print(fruits)
fruits=["apple", "banana", "cherry"]
fruits1=["mango", "pineapple", "papaya"]
fruits.extend(fruits1)
print(fruits)
#remove list items
list = ["apple", "banana", "cherry"]
list.remove("banana")
print(list)
list = ["apple", "banana", "cherry"]
list.pop(2)
print(list)
list = ["apple", "banana", "cherry"]
del list[0]
print(list)
list = ["apple", "banana", "cherry"]
list.clear()
print(list)
```

```python
#loop in list
list = ["apple", "banana", "cherry"]
for x in list:
    print(x)
#using while
list = ["apple", "banana", "cherry"]
i = 0
while i<len(list):
    print(list[i])
    i = i + 1
#sort
list = ["orange", "mango", "kiwi", "pineapple", "banana"]
list.sort()
print(list)
#reverse
list = ["orange", "mango", "kiwi", "pineapple", "banana"]
list.reverse()
print(list)
#copy list
list = ["apple", "banana", "cherry"]
list1 = list.copy()
print(list1)
#join lists
list1=["a", "b", "c"]
list2=["apple"]
print(list1+list2)
#joining lists using forloop
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]
for x in list2:
    list1.append(x)
    print(list1)
#using extend method
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]
list1.extend(list2)
print(list1)
```
