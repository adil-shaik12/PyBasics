```python
def binarySearch(arr,val):
    left=0
    right=len(arr)-1
    while left<=right:
        mid=(left+right)//2
        if arr[mid]==val:
            return mid
        if arr[mid]<val:
            left=mid+1
        else:
            right=mid-1
    return -1
list=[1,3,4,6,7,9,10,12,14,15,16]
x=9
result=binarySearch(list,x)
if result!=-1:
    print("found at index",result)
else:
    print("not found")
```
