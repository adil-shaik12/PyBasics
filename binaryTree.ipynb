```python
class node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
root=node(5)
nodea=node(4)
nodeb=node(7)
nodec=node(6)
noded=node(3)
nodee=node(2)
nodef=node(9)
root.left=nodea
root.right=nodeb
nodea.left=nodec
nodea.right=noded
nodeb.left=nodee
nodeb.right=nodef
print("root.data",root.data)
print("root.left.data",root.left.data)
print("root.right.data",root.right.data)
print("root.left.left.data",root.left.left.data)
print("root.left.right.data",root.left.right.data)
print("root.right.left.data",root.right.left.data)
print("root.right.right.data",root.right.right.data)
def inorder(node):
    if node is None:
        return
    inorder(node.left)
    print(node.data,end=",")
    inorder(node.right)
print("inorder traversal:")
inorder(root)
print()
def preorder(node):
    if node is None:
        return
    print(node.data,end=",")
    preorder(node.left)
    preorder(node.right)
print("preorder traversal")
preorder(root)
print()
def postorder(node):
    if node is None:
        return
    postorder(node.left)
    postorder(node.right)
    print(node.data,end=",")
print("postorder traversal")
postorder(root) 
print()

#search for a node
def search(node,val):
    if node is None:
        return None
    if node.data==val:
        return node
    left=search(node.left,val)
    if left:
        return left
    return search(node.right,val)
result=search(root,9)
if result:
    print(F"key found in a tree:{result.data}")
else:
    print("key not found in a tree")
    
#to insert a node
def insert(root,val):
    if root is None:
        return node(val)
    queue=[root]
    while queue:
        temp=queue.pop(0)
        if temp.left is None:
            temp.left=node(val)
            return root
        else:
            queue.append(temp.left)
        if temp.right is None:
            temp.right=node(val)
            return root
        else:
            queue.append(temp.right)
    return root
print("after inserton:")
node=insert(root,8)
inorder(root)
print()

#to find a low value
def minValue(node):
    if node is None:    
        return float('inf')
    left=minValue(node.left)
    right=minValue(node.right)
    return min(node.data,left,right)
print("minimum value is:",minValue(root))

#to delete a node
def deldeepest(root,key):
    queue=[root]
    while queue:
        temp=queue.pop(0)
        if temp.left:
            if temp.left == key:
                temp.left=None
                return
            else:
                queue.append(temp.left)
        if temp.right:
            if temp.right == key:
                temp.right=None
                return
            else:
                queue.append(temp.right)
def delete(root,data):
    if root is None:
        return None
    if root.left is None and root.right is None:
        if root.data==data:
            return None
        else:
            return root
    keynode=None
    queue=[root]
    while queue:
        temp=queue.pop(0)
        if temp.data==data:
            keynode=temp
        if temp.left:
            queue.append(temp.left)
        if temp.right:
            queue.append(temp.right)
        deepest=temp
    if keynode is None:
        return root
    keynode.data=deepest.data
    deldeepest(root,deepest)
    return root
print("before deleting:")
inorder(root)
print()
root=delete(root,6)
print("after deleting")  
inorder(root)
        
    
```
